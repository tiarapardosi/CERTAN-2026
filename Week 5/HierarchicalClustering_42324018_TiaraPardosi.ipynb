```python
import numpy as np
import pandas as pd
from scipy.cluster.hierarchy import linkage, dendrogram
from scipy.spatial.distance import pdist, squareform
import matplotlib.pyplot as plt
# 1. Dataset sederhana
data = np.array([
 [1.0, 1.0], # a
 [1.5, 1.5], # b
 [5.0, 5.0], # c
 [3.0, 4.0], # d
 [4.0, 4.0], # e
 [3.0, 3.5] # f
])
# 2. Hitung Distance Matrix (Euclidean Distance)
distance_matrix = pdist(data, metric='euclidean')
distance_df = pd.DataFrame(squareform(distance_matrix),
 columns=['a', 'b', 'c', 'd', 'e', 'f'],
 index=['a', 'b', 'c', 'd', 'e', 'f'])
# 3. Tampilkan Distance Matrix
print("=== Distance Matrix (Euclidean) ===")
print(distance_df)
# 4. Hierarchical Clustering (Single Linkage)
linked = linkage(data, method='single')
# 5. Plot Dendrogram
plt.figure(figsize=(12, 8))
plt.title("Hierarchical Clustering Dendrogram (Single Linkage)")
dendrogram(linked, labels=['a', 'b', 'c', 'd', 'e', 'f'], leaf_rotation=90)
plt.xlabel("Data Points")
plt.ylabel("Distance")
plt.grid(True)
plt.show()
```

# Perbandingan Hierarchical Clustering dengan Single, Complete, dan Average Linkage

Pada percobaan ini dilakukan perbandingan hasil Hierarchical Clustering menggunakan tiga metode linkage, yaitu:
1. Single Linkage
2. Complete Linkage
3. Average Linkage

Dataset yang digunakan merupakan dataset sederhana dengan enam titik data yaitu a, b, c, d, e, dan f.

1. Import Library

Library yang digunakan adalah NumPy untuk mengolah data, Matplotlib untuk membuat visualisasi dendrogram, serta fungsi `linkage` dan `dendrogram` dari SciPy untuk melakukan Hierarchical Clustering.

## 2. Dataset

Dataset terdiri dari enam titik data dua dimensi. Setiap titik diberi label a sampai f.

Data:
- a = (1.0, 1.0)
- b = (1.5, 1.5)
- c = (5.0, 5.0)
- d = (3.0, 4.0)
- e = (4.0, 4.0)
- f = (3.0, 3.5)

Dataset ini sama dengan dataset yang digunakan pada percobaan pertama dalam modul.

## 3. Hierarchical Clustering dengan Single Linkage

Single Linkage menentukan jarak antara dua cluster berdasarkan jarak terpendek antara anggota dari kedua cluster.

Metode ini cenderung menggabungkan titik yang paling dekat terlebih dahulu sehingga dapat menghasilkan efek chaining, yaitu beberapa titik tergabung secara bertahap karena memiliki anggota yang berdekatan.

## 4. Hierarchical Clustering dengan Complete Linkage

Complete Linkage menentukan jarak antara dua cluster berdasarkan jarak terjauh antara anggota dari kedua cluster.

Dengan menggunakan jarak terjauh, metode ini cenderung menghasilkan cluster yang lebih kompak dibandingkan Single Linkage.

## 5. Hierarchical Clustering dengan Average Linkage

Average Linkage menentukan jarak antara dua cluster berdasarkan rata-rata jarak seluruh pasangan titik dari kedua cluster.

Metode ini berada di antara Single Linkage dan Complete Linkage karena tidak hanya mempertimbangkan jarak terdekat atau terjauh, tetapi menggunakan rata-rata jarak antar anggota cluster.

## 6. Perbandingan Ketiga Metode Linkage

Untuk mempermudah perbandingan, ketiga dendrogram ditampilkan secara bersamaan. Dengan demikian, dapat dilihat perbedaan proses penggabungan cluster dan ketinggian setiap penggabungan.

```python
import pandas as pd
df = pd.read_csv("Mall_Customers.csv")
df.head()
```

```python
x = df.iloc[:, [3, 4]].values
```

```python
from sklearn.preprocessing import StandardScaler
data_scaler = StandardScaler()
scaled_data = data_scaler.fit_transform(x)
scaled_data.shape
```

```python

from scipy.cluster.hierarchy import linkage, dendrogram
clustering = linkage(scaled_data, method="complete", metric="euclidean")
```

```python
import matplotlib.pyplot as plt
dendrogram(clustering)
plt.show()
```

## 7. Analisis Hasil

Berdasarkan hasil dendrogram, terdapat perbedaan cara ketiga metode dalam menggabungkan data.

### Single Linkage
Single Linkage menggunakan jarak terdekat antar anggota cluster. Pada dataset ini, penggabungan cluster dapat terjadi secara bertahap berdasarkan pasangan titik yang memiliki jarak paling dekat. Metode ini lebih mudah menghasilkan efek chaining.

### Complete Linkage
Complete Linkage menggunakan jarak terjauh antar anggota cluster. Oleh karena itu, penggabungan cluster dilakukan dengan mempertimbangkan agar anggota dalam cluster tidak memiliki jarak yang terlalu jauh. Hasilnya cenderung membentuk cluster yang lebih kompak.

### Average Linkage
Average Linkage menggunakan rata-rata jarak antar anggota dari dua cluster. Hasilnya menjadi pendekatan di antara Single Linkage dan Complete Linkage karena tidak hanya bergantung pada jarak terdekat maupun jarak terjauh.

Dengan demikian, ketiga metode dapat menghasilkan struktur dendrogram yang berbeda meskipun menggunakan dataset yang sama.

TUGAS

```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.cluster.hierarchy import linkage, dendrogram

# Dataset sederhana
data = np.array([
    [1.0, 1.0],  # a
    [1.5, 1.5],  # b
    [5.0, 5.0],  # c
    [3.0, 4.0],  # d
    [4.0, 4.0],  # e
    [3.0, 3.5]   # f
])

labels = ['a', 'b', 'c', 'd', 'e', 'f']

# Single Linkage
single = linkage(data, method='single', metric='euclidean')

# Complete Linkage
complete = linkage(data, method='complete', metric='euclidean')

# Average Linkage
average = linkage(data, method='average', metric='euclidean')

# Menampilkan ketiga dendrogram
plt.figure(figsize=(18, 5))

plt.subplot(1, 3, 1)
dendrogram(single, labels=labels)
plt.title("Single Linkage")
plt.xlabel("Data Points")
plt.ylabel("Distance")
plt.grid(True)

plt.subplot(1, 3, 2)
dendrogram(complete, labels=labels)
plt.title("Complete Linkage")
plt.xlabel("Data Points")
plt.ylabel("Distance")
plt.grid(True)

plt.subplot(1, 3, 3)
dendrogram(average, labels=labels)
plt.title("Average Linkage")
plt.xlabel("Data Points")
plt.ylabel("Distance")
plt.grid(True)

plt.tight_layout()
plt.show()
```

## Hasil Perbandingan

Berdasarkan ketiga dendrogram, terdapat perbedaan pada urutan penggabungan cluster dan jarak penggabungannya. Single Linkage menggunakan jarak terdekat antar titik sehingga cenderung membentuk cluster secara bertahap. Complete Linkage menggunakan jarak terjauh sehingga menghasilkan cluster yang lebih kompak. Sedangkan Average Linkage menggunakan rata-rata jarak antar anggota cluster sehingga hasilnya berada di antara Single Linkage dan Complete Linkage.
