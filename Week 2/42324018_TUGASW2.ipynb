```python
## Analisis Previous Purchases

import pandas as pd

path_shopping = r"C:\Perkuliahan\Semester 5\CERTAN\Week 1\TUGAS CERTAN\CERTAN-2026\Week 2\shopping_trends.csv"

shopping = pd.read_csv(path_shopping)

shopping.head()
```

```python
print("Mean :", shopping['Previous Purchases'].mean())
print("Std  :", shopping['Previous Purchases'].std())
print("Min  :", shopping['Previous Purchases'].min())
print("Q1   :", shopping['Previous Purchases'].quantile(0.25))
print("Q2   :", shopping['Previous Purchases'].quantile(0.50))
print("Q3   :", shopping['Previous Purchases'].quantile(0.75))
print("Max  :", shopping['Previous Purchases'].max())
```

Hasil Analisis

Berdasarkan hasil statistik deskriptif, jumlah Previous Purchases memiliki rata-rata sebesar 25,35 pembelian. Nilai minimum adalah 1 pembelian, sedangkan nilai maksimum mencapai 50 pembelian. Nilai median (Q2) sebesar 25, yang menunjukkan bahwa sekitar setengah pelanggan memiliki jumlah pembelian sebelumnya di bawah atau sama dengan 25 dan setengah lainnya di atas atau sama dengan 25. Nilai Q1 sebesar 13 dan Q3 sebesar 38, sehingga sebagian besar data berada pada rentang 13 sampai 38 pembelian sebelumnya

```python
##Analisis korelasi Age dengan Purchase Amount

shopping[['Age', 'Purchase Amount (USD)']].corr()
```

Hasil Analisis

Korelasi antara Age dan Purchase Amount (USD) digunakan untuk mengetahui apakah usia pelanggan memiliki hubungan dengan jumlah pembelian yang dilakukan. Jika nilai korelasi positif, maka peningkatan usia cenderung diikuti peningkatan jumlah pembelian. Sebaliknya, jika bernilai negatif, peningkatan usia cenderung diikuti penurunan jumlah pembelian. Semakin mendekati 0, semakin lemah hubungan antara kedua variabel.

```python
## Analisis korelasi Purchase Amount dengan Review Rating

shopping[['Purchase Amount (USD)', 'Review Rating']].corr()
```

Hasil Analisis

Korelasi antara Purchase Amount (USD) dan Review Rating digunakan untuk melihat hubungan antara jumlah pembelian dengan penilaian yang diberikan pelanggan. Nilai korelasi positif menunjukkan bahwa jumlah pembelian yang lebih tinggi cenderung berkaitan dengan rating yang lebih tinggi, sedangkan nilai negatif menunjukkan kecenderungan hubungan yang berlawanan. Nilai yang mendekati 0 menunjukkan bahwa hubungan antara jumlah pembelian dan rating relatif lemah.
